```python
with open("One.txt") as mytext:
    words_one = mytext.read().lower().split()
```

```python
uni_words_one = set(words_one)
```

```python
with open("Two.txt") as mytext:
    words_two = mytext.read().lower().split()
```

```python
uni_words_two = set(words_two)
```

```python
uni_words_two
```

```python
all_uni_words = set()
all_uni_words.update(uni_words_one)
all_uni_words.update(uni_words_two)
```

```python
all_uni_words
```

```python
full_vocab = dict()
i = 0

for word in all_uni_words:
    full_vocab[word] = i
    i += 1
```

```python
full_vocab
```

```python
one_freq = [0] * len(full_vocab)
two_freq = [0] * len(full_vocab)
all_words = [""] * len(full_vocab)
```

```python
one_freq
```

```python
two_freq
```

```python
all_words
```

```python
for word in full_vocab:
    word_ind = full_vocab[word]
    all_words[word_ind] = word
```

```python
all_words
```

```python
with open('One.txt') as f:
    one_text = f.read().lower().split()

for word in one_text:
    word_ind = full_vocab[word]
    one_freq[word_ind] += 1
```

```python
one_freq
```

```python
with open('Two.txt') as f:
    two_text = f.read().lower().split()
for word in two_text:
    word_ind = full_vocab[word]
    two_freq[word_ind] += 1
```

```python
two_freq
```

```python
import pandas as pd
pd.DataFrame(data = [one_freq, two_freq], columns = all_words)
```

```python
text = ['This is a line',
    "This is another line",
    "Completely different line"]
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, TfidfTransformer
```

```python
cv = CountVectorizer()
```

```python
cv.fit_transform(text)
```

```python
sparse_mat = cv.fit_transform(text)
```

```python
sparse_mat.todense()
```

```python
cv.vocabulary_
```

```python
cv = CountVectorizer(stop_words='english')
```

```python
cv.fit_transform(text).todense()
```

```python
cv.vocabulary_
```

```python
tfidf_transformer = TfidfTransformer()
```

```python
cv = CountVectorizer()
```

```python
counts = cv.fit_transform(text)
```

```python
counts
```

```python
tfidf = tfidf_transformer.fit_transform(counts)
```

```python
tfidf.todense()
```

```python
from sklearn.pipeline import Pipeline
```

```python
pipe = Pipeline([("cv", CountVectorizer()), ("tfidf", TfidfTransformer())])
```

```python
results = pipe.fit_transform(text)
```
